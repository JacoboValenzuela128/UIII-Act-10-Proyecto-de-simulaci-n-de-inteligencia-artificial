# JUEGO DE ADIVINANZA

# Prompt
Me puedes generar esto? Juego de adivinanza de palabra: el usuario intenta adivinar una palabra letra por letra. Muestra progreso (_ A _ _) y permite 10 intentos.
no funciones, no programación orientada a objetos, no listas, no diccionarios, no try, solamente lo que pide el problema.

```python
# Palabra a adivinar y su longitud
palabra_secreta = "PYTHON"
longitud = 6

# Variables para registrar el progreso del jugador
progreso = "______"
intentos_restantes = 10

print("¡Bienvenido al juego de adivinanza de palabras!")
print("La palabra tiene", longitud, "letras.")
print("Progreso actual:", progreso)
print("Tienes", intentos_restantes, "intentos.")
print("-----------------------------------")

# Ciclo principal del juego
while intentos_restantes > 0 and progreso != palabra_secreta:
    # Pedir letra al usuario
    letra_usuario = input("Introduce una letra (en mayúscula): ")

    # Variable para saber si la letra estaba en la palabra
    letra_acertada = False
    nuevo_progreso = ""

    # Verificar la letra posición por posición de forma manual
    # Posición 0
    if palabra_secreta[0] == letra_usuario:
        nuevo_progreso = nuevo_progreso + letra_usuario
        letra_acertada = True
    else:
        nuevo_progreso = nuevo_progreso + progreso[0]

    # Posición 1
    if palabra_secreta[1] == letra_usuario:
        nuevo_progreso = nuevo_progreso + letra_usuario
        letra_acertada = True
    else:
        nuevo_progreso = nuevo_progreso + progreso[1]

    # Posición 2
    if palabra_secreta[2] == letra_usuario:
        nuevo_progreso = nuevo_progreso + letra_usuario
        letra_acertada = True
    else:
        nuevo_progreso = nuevo_progreso + progreso[2]

    # Posición 3
    if palabra_secreta[3] == letra_usuario:
        nuevo_progreso = nuevo_progreso + letra_usuario
        letra_acertada = True
    else:
        nuevo_progreso = nuevo_progreso + progreso[3]

    # Posición 4
    if palabra_secreta[4] == letra_usuario:
        nuevo_progreso = nuevo_progreso + letra_usuario
        letra_acertada = True
    else:
        nuevo_progreso = nuevo_progreso + progreso[4]

    # Posición 5
    if palabra_secreta[5] == letra_usuario:
        nuevo_progreso = nuevo_progreso + letra_usuario
        letra_acertada = True
    else:
        nuevo_progreso = nuevo_progreso + progreso[5]

    # Actualizar el progreso con los cambios de esta ronda
    progreso = nuevo_progreso

    # Lógica de intentos e impresión de resultados
    if letra_acertada == True:
        print("¡Bien hecho! La letra está en la palabra.")
    else:
        intentos_restantes = intentos_restantes - 1
        print("Incorrecto. Perdiste un intento.")

    print("Progreso actual:", progreso)
    print("Intentos restantes:", intentos_restantes)
    print("-----------------------------------")

# Pantalla final del juego
if progreso == palabra_secreta:
    print("¡Felicidades! Ganaste. Adivinaste la palabra:", palabra_secreta)
else:
    print("Te has quedado sin intentos. Perdiste. La palabra era:", palabra_secreta)
```
