```python
# ==========================================================
# SIMULADOR IA: MAKEUP STUDIO MANAGER (Lista de Diccionarios)
# ==========================================================

print("==================================================")
print("   ¡BIENVENIDO A AI MAKEUP STUDIO MANAGER v4.0!   ")
print("  Gestiona tu equipo de estilistas con IA y Datos ")
print("==================================================")

# Estructuras solicitadas vacías al inicio
lista_maquillistas = []
losmaquillistas = {}

while True:
    # Impresión del Menú Principal
    print("\n--- MENÚ DE GESTIÓN INTELIGENTE ---")
    print("1. Registrar Maquillista y asignar especialidad")
    print("2. Mostrar equipo de Maquillistas (Escanear Lista con For)")
    print("3. Contador de Maquillistas por Nivel")
    print("4. Cerrar Sistema (Salir)")

    opcion = input("\nSelecciona una opción (1-4): ")

    # ------------------------------------------------------
    # OPCIÓN 1: REGISTRAR MAQUILLISTA (Crear Diccionario y guardar en Lista)
    # ------------------------------------------------------
    if opcion == "1":
        print("\n--- REGISTRO DE NUEVO TALENTO ---")
        nombre = input("Nombre del maquillista: ")
        anos_experiencia = input("¿Cuántos años de experiencia tiene?: ")
        anos_experiencia = int(anos_experiencia) # Convertimos a entero

        # Clasificación automática de la IA según experiencia
        if anos_experiencia < 2:
            nivel_ia = "Junior (Principiante)"
            especialidad = "Maquillaje Social / Diario"
        elif anos_experiencia >= 2 and anos_experiencia <= 5:
            nivel_ia = "Advanced (Avanzado)"
            especialidad = "Maquillaje de Novias / Eventos"
        else:
            nivel_ia = "Master (Experto)"
            especialidad = "Efectos Especiales (FX) / Pasarelas"

        # Creamos el diccionario con los datos del maquillista actual
        losmaquillistas = {
            "nombre": nombre,
            "experiencia": anos_experiencia,
            "rango": nivel_ia,
            "especialidad": especialidad
        }

        # Guardamos el diccionario dentro de la lista
        lista_maquillistas.append(losmaquillistas)

        print(f"\n🤖 IA: ¡{nombre} ha sido registrado exitosamente!")
        print(f"🤖 IA Clasificación: Nivel {nivel_ia} -> Especialidad en {especialidad}.")

    # ------------------------------------------------------
    # OPCIÓN 2: MOSTRAR EQUIPO (Uso de ciclo FOR para recorrer la lista)
    # ------------------------------------------------------
    elif opcion == "2":
        print("\n--- ESCANEANDO BASE DE DATOS CON IA ---")

        if lista_maquillistas == []:
            print("❌ La lista está vacía. No hay maquillistas registrados aún.")
        else:
            print("✅ Personal Activo en el Studio:")
            print("--------------------------------------------------")

            # Ciclo FOR para recorrer cada diccionario dentro de la lista
            for maquillista in lista_maquillistas:
                print(f"• Maquillista: {maquillista['nombre']}")
                print(f"  Experiencia: {maquillista['experiencia']} años")
                print(f"  Rango IA:    {maquillista['rango']}")
                print(f"  Enfocado en: {maquillista['especialidad']}")
                print("--------------------------------------------------")

    # ------------------------------------------------------
    # OPCIÓN 3: ESTADÍSTICAS E INTELIGENCIA DE DATOS
    # ------------------------------------------------------
    elif opcion == "3":
        print("\n--- ANALÍTICA DE IA ---")

        if lista_maquillistas == []:
            print("❌ No hay datos suficientes para generar estadísticas.")
        else:
            # Variables contadoras
            total_empleados = len(lista_maquillistas)
            expertos_master = 0

            # Usamos FOR e IF para contar cuántos Master FX tenemos
            for m in lista_maquillistas:
                if m["rango"] == "Master (Experto)":
                    expertos_master = expertos_master + 1

            print(f"-> Total de maquillistas en el sistema: {total_empleados}")
            print(f"-> Maquillistas nivel MASTER FX: {expertos_master}")

            # Sugerencia de la IA basada en datos
            if expertos_master == 0:
                print("💡 Consejo de la IA: Tu equipo es muy joven. Deberías contratar a un Master para pasarelas.")
            else:
                print("💡 Consejo de la IA: ¡Excelente balance! Tienes talento calificado para eventos de alta costura.")

    # ------------------------------------------------------
    # OPCIÓN 4: SALIR DEL PROGRAMA
    # ------------------------------------------------------
    elif opcion == "4":
        print("\nCerrando sistema de IA... Datos respaldados. ¡Hasta pronto!")
        break

    # CONTROL DE ERRORES DEL MENÚ
    else:
        print("❌ Opción inválida. Digita un número del 1 al 4.")

    print("\n==================================================")
    print("Programa realizado por Arath Valenzuela 0148 NL57")
```
